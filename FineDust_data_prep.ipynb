한글이 깨지는 파일을 올바른 인코딩으로 읽어오기

```python
import pandas as pd

# 한글이 깨지는 파일을 올바른 인코딩으로 읽어오기 (예: EUC-KR, CP949)
input_file = '/content/drive/MyDrive/서울특별시_시간별 (초)미세먼지_20221231/서울시 대기질 자료 제공_2020-2021.csv'  # 원본 파일 경로
output_file = '/content/drive/MyDrive/서울특별시_시간별 (초)미세먼지_20221231/서울시 대기질 자료 제공_2020-2021.csv'  # 저장할 파일 경로

# 올바른 인코딩으로 파일을 읽습니다.
df = pd.read_csv(input_file, encoding='euc-kr')  # or 'cp949' if that works

# UTF-8로 인코딩을 변경하여 저장합니다.
df.to_csv(output_file, encoding='utf-8', index=False)

print("인코딩 변환이 완료되었습니다. UTF-8로 저장되었습니다.")

```

데이터를 다 합치고 시간 순으로 정리한 데이터를 엑셀로 다운로드


```python
import pandas as pd

# File paths for all 4 Excel files
file_paths = [
    '/content/filtered_2012-2015.xlsx',
    '/content/filtered_2016-2019.xlsx',
    '/content/filtered_2020-2021.xlsx',
    '/content/filtered_2022.xlsx'
]

# Load all Excel files into pandas DataFrames
dfs = [pd.read_excel(file) for file in file_paths]

# Concatenate all the DataFrames into a single DataFrame
merged_df = pd.concat(dfs, ignore_index=True)

# Convert the 'Date' column to datetime format, letting Pandas infer the format automatically
# Replace 'Date' with the actual column name if necessary
merged_df['Date'] = pd.to_datetime(merged_df['Date'], infer_datetime_format=True)

# Sort the DataFrame by the 'Date' column in ascending order
sorted_df = merged_df.sort_values(by='Date', ascending=True)

# Save the merged and sorted DataFrame to a new Excel file
output_file = 'Transformed_data.xlsx'
sorted_df.to_excel(output_file, index=False)

print(f"Merged and sorted data has been saved to {output_file}.")
```
